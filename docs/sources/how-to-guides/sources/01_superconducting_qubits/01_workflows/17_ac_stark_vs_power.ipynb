## Background
In this how-to guide, you'll perform a measurement to find the qubit frequency using an `ac_stark_vs_power` experiment included in the LabOne Q Applications Library. In an AC Stark shift experiment, you'll vary the off-resonant drive-pulse amplitude, $\Omega$, and do a qubit spectrosocpy simultaneously.

![](../../../images/ac_stark.svg "Demonstration of pulse sequence.")

The qubit frequency $\omega$ changes with the drive amplitude $\Omega_d$ in the following way:

$$\Delta\omega_\text{AC} \approx -2\chi n, \quad n\sim \Omega_d^2$$

where $\chi$ is the chi-shift and $n$ the resonator photon number.

This results in data that will, ignoring any other effects, look like:

![](<../../../images/ac_stark_shift.png> "Plot of analyzer result of the ac stark shift experiment.")

The experiment workflow also accepts a list of qubits, in which case the AC Stark shift of all the qubits is measured in parallel within a single experiment. Note, however, that this is only possible for qubits whose readout lines are connected to *different physical QA channels*: the experiment sweeps the amplitude of the readout line and uses its hardware oscillator, which are shared by all the qubits multiplexed on the same QA channel. The workflow verifies this condition before creating the experiment and raises an error if two of the passed qubits share a QA channel. Qubits multiplexed on the same readout line must instead be measured in separate workflow runs.

### DeviceSetup

This guide requires a setup that can drive and readout tunable transmon qubits. Your setup could contain an SHFQC+ instrument, or an SHFSG and an SHFQA instruments. Here, we will use an SHFQC+ with 6 signal generation channels and a PQSC.

If you have used LabOne Q before and already have a `DeviceSetup` for your setup, you can reuse that.

If you do not have a `DeviceSetup`, you can create one using the code below. Just change the device numbers to the ones in your rack and adjust any other input parameters as needed.

````python
from laboneq.contrib.example_helpers.generate_descriptor import generate_descriptor

descriptor = generate_descriptor(
    pqsc=["DEV10001"],
    shfqc_6=["DEV12001"],
    number_data_qubits=number_of_qubits,
    multiplex=True,
    number_multiplex=number_of_qubits,
    include_cr_lines=False,
    ip_address="localhost",
)
setup = DeviceSetup.from_descriptor(descriptor, "localhost")
````

### Qubits

We will generate a single `TunableTransmonQubit` from the logical signal groups in our `DeviceSetup`. The name of the logical signal group, `q0`, will be the UIDs of the qubit. Moreover, the qubit will have the same logical signal line as the ones of the logical signal group in the `DeviceSetup`.

````python
from laboneq_applications.qpu_types.tunable_transmon import TunableTransmonQubit

qubits = TunableTransmonQubit.from_device_setup(setup)
````

````python
for q in qubits:
    print("-------------")
    print("Qubit UID:", q.uid)
    print("Qubit logical signals:")
    for sig, lsg in q.signals.items():
        print(f"  {sig:<10} ('{lsg:>10}')")
````

Configure the qubit parameters to reflect the properties of the qubits on your QPU using the following code:

````python
for q in qubits:
    q.parameters.ge_drive_pulse["sigma"] = 0.25
    q.parameters.readout_amplitude = 0.5
    q.parameters.reset_delay_length = 1e-6
    q.parameters.readout_range_out = -25
    q.parameters.readout_lo_frequency = 7.4e9

qubits[0].parameters.drive_lo_frequency = 6.4e9
qubits[0].parameters.resonance_frequency_ge = 6.3e9
qubits[0].parameters.resonance_frequency_ef = 6.0e9
qubits[0].parameters.readout_resonator_frequency = 7.0e9
````

### Quantum Operations

Create the set of `TunableTransmonOperations`:

````python
from laboneq_applications.qpu_types.tunable_transmon import TunableTransmonOperations

qops = TunableTransmonOperations()
````

### QPU

Create the `QPU` object from the qubits and the quantum operations

````python
from laboneq.dsl.quantum import QPU

qpu = QPU(qubits, quantum_operations=qops)
````

## Running the Experiment Workflow

You'll now instantiate the experiment workflow and run it. For more details on what experiment workflows are and what tasks they execute, see the [Experiment Workflows tutorial](../../../../tutorials/sources/experiment_workflows.ipynb).

You'll start by importing `numpy`, the AC Stark shift experiment workflow from `laboneq_applications`, as well as `plot_simulation` for inspecting the experiment sequence.

````python
import numpy as np
from laboneq.contrib.example_helpers.plotting.plot_helpers import plot_simulation

from laboneq_applications.contrib.experiments import ac_stark_shift_vs_power
````

Let's first create the options class for the AC Stark shift experiment and inspect it using the `show_fields` function from the `workflow` namespace of LabOne Q, which was imported from `laboneq.simple`:

````python
options = ac_stark_shift_vs_power.experiment_workflow.options()
options.count(2**9)
options.do_fitting(True)
workflow.show_fields(options)
````

Notice that:

- the experiment is run in `AcquisitionType.SPECTROSCOPY`.
- the analysis workflow will run automatically (`do_analysis=True`)
- the figures produced by the analysis are automatically closed (`close_figures=True`)
- the qubit parameters will not be updated (`update=False`)

Here, let's disable closing the figures produced by the analysis so we see them in the cell output. Note however that the fit attempted by the analysis routine in emulation mode will not be representative, because we do not acquire data from a real experiment.

````python
options.close_figures(False)
````

Now we run the experiment workflow on the first qubit.

Note that the fit results in emulation mode are not meaningful, because we do not acquire data from a real experiment.

````python
# our qubits live here in the demo setup:
qubits = qpu.quantum_elements

center_q1 = qubits[0].parameters.resonance_frequency_ge

off = 100e6

exp_workflow = ac_stark_shift_vs_power.experiment_workflow(
    session=session,
    qpu=qpu,
    qubits=qubits[0].uid,
    amplitudes=np.geomspace(0.2, 1, 21),
    frequencies=np.linspace(center_q1 - off, center_q1 + off, 151),
    options=options,
)

workflow_results = exp_workflow.run()
````

### Inspect the Tasks That Were Run

````python
for t in workflow_results.tasks:
    print(t)
````

### Inspect the Output Simulation

You can also inspect the compiled experiment and plot the simulated output:

````python
compiled_experiment = workflow_results.tasks["compile_experiment"].output
plot_simulation(compiled_experiment, length=1000e-6)
````
